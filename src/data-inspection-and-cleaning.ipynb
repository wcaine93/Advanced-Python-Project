### 5. Check the empty value.

```python
df.isnull().sum()
```

The empty value is in column `Water`. Show row:

```python
df[df['Water'].isnull()]
```

### 6. Replace Empty Values Using the value that appears most frequently.

```python
df['Water'] = df['Water'].fillna(df['Water'].mode()[0])
```

### 7. Check and Discovering Duplicates.

```python
df[df.duplicated()]
```

### 8. Removing Duplicates.

```python
df.drop_duplicates(inplace=True)
```
