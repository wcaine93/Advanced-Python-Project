# B: Split the data into training and testing data

```python
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split

df = pd.read_csv('clean_data.csv')
```

### 1. Create the data features (X) and target labels (y).  

```python
X = df.drop('Strength', axis=1)
y = df['Strength']
```

### 2. Randomly split the data features and labels into a training and test sets by holding 30% of the data for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# F: Increase the number of hidden layers

### 1. Four hidden layers, each of 10 nodes and ReLU activation function.



```python
increased_layers_model = tf.keras.models.Sequential([
  tf.keras.Input(shape=(8,)),
  tf.keras.layers.Dense(10, activation='relu'),
  tf.keras.layers.Dense(10, activation='relu'),
  tf.keras.layers.Dense(10, activation='relu'),
  tf.keras.layers.Dense(10, activation='relu'),
  tf.keras.layers.Dense(1)
])
```

```python
increased_layers_model.build()
```

### 2. Plot the model architecture

```python
tf.keras.utils.plot_model(increased_layers_model, to_file='increased_layers_model.png', show_shapes=True)
```

### 3. Compile the model using the adam as the optimizer and using the mean squared error as the loss function.

```python
increased_layers_model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
```

### 4. Train the model on the training data using 100 epochs.

```python
increased_layers_model.fit(
    X_train, y_train,
    epochs=100
)
```

### 5. Evaluate the model on the test data using model.evaluate

```python
increased_layers_model.evaluate(X_test, y_test)
```

Ultimate MSE of `94.966`

### 6. Make predictions on test data using model.predict

```python
increased_layers_model.predict(X_test)
```
